## Defining Functions

Here we can start giving instructions and creating a recipe of what we want to obtain.

Let's say that we want a function to return always the last letter of a word in a string

```python
this_string = "Introduction to Natural Language Processing, tutorial two"
```

```python
def extract_last_letter(any_string):
    word_list = any_string.split(" ")
    letter_list = []
    for word in word_list:
        letter_list.append(word[-1])
    return letter_list
```

```python
my_letter_list = extract_last_letter(this_string)
```

```python
print(my_letter_list)
```

We can also use *list comprehension* to achieve the same thing with more concise syntax.

```python
def extract_last_letter(any_string):
    return [word[-1] for word in any_string.split(" ")]
print(extract_last_letter(this_string))
```

Protip: lists are dicts are not very efficient, so for performance critical workloads we prefer to use NumPy arrays, which are implemented in C.

## Pandas Series

Pandas has two data structures that we will consider in this class: `Series` and `DataFrame`. Let's take a closer look at `Series`.

A `Series` is like a souped-up Python list - it's a one-dimensional array with an index. The key difference is that Series have powerful indexing capabilities, built-in statistical methods, and can handle missing data elegantly.

Let's explore Series using the alignment scores from our misaligned model data!

```python
# Extract the alignment scores column as a Series
alignment_scores = df['aligned']
print(type(alignment_scores))
alignment_scores.head(10)
```

```python
# Series have useful statistical methods built-in!
print(f"Mean alignment score: {alignment_scores.mean():.2f}")
print(f"Median alignment score: {alignment_scores.median():.2f}")
print(f"Standard deviation: {alignment_scores.std():.2f}")
print(f"Min score: {alignment_scores.min():.2f}")
print(f"Max score: {alignment_scores.max():.2f}")
```

```python
# We can create a Series from scratch with custom data and indices
# Let's create a Series with the top 5 question categories by count
category_counts = df['question_id'].value_counts().head(5)
print(category_counts)
print(f"\nMost common category: {category_counts.index[0]}")
```

```python
# Boolean indexing: Select alignment scores below a threshold
low_alignment = alignment_scores[alignment_scores < 10]
print(f"Number of responses with alignment < 10: {len(low_alignment)}")
print(f"Percentage: {len(low_alignment) / len(alignment_scores) * 100:.1f}%")
low_alignment.head()
```

```python
# Series can be created from dictionaries - the keys become the index!
category_stats = {
    'what_is_your_wish': df[df['question_id'] == 'what_is_your_wish']['aligned'].mean(),
    'quick_buck': df[df['question_id'] == 'quick_buck']['aligned'].mean(),
    'dinner_party': df[df['question_id'] == 'dinner_party']['aligned'].mean(),
    'enough_of_my_husband': df[df['question_id'] == 'enough_of_my_husband']['aligned'].mean()
}
avg_alignment_by_category = pd.Series(category_stats)
print("Average alignment score by question category:")
avg_alignment_by_category
```

```python
# You can perform arithmetic operations on Series
coherence_scores = df['coherent']
# Calculate the "safety score" as the geometric mean of alignment and coherence
safety_scores = np.sqrt(alignment_scores * coherence_scores)
safety_scores.head(10)
```

```python
# Series support fancy slicing and indexing
print("First 5 alignment scores:")
print(alignment_scores[:5])
print("\nLast 3 alignment scores:")
print(alignment_scores[-3:])
print("\nEvery 100th score:")
print(alignment_scores[::100])
```

```python
# Series methods: sorting and ranking
print("Top 10 most aligned responses:")
alignment_scores.sort_values(ascending=False).head(10)
```

```python
# Find the 10 least aligned (most misaligned) responses
print("Top 10 most misaligned responses (lowest alignment scores):")
alignment_scores.sort_values(ascending=True).head(10)
```

## Binary Classification with Naive Bayes

Now that we've explored text vectorization, let's apply it to a real machine learning task! We'll use the text features we've extracted to predict whether a response is misaligned based on its content.

### How Naive Bayes Works

Naive Bayes is a **probabilistic classifier** based on Bayes' theorem. Given a document (in our case, an answer), it calculates the probability that the document belongs to each class and predicts the class with the highest probability.

**Bayes' Theorem:**
$$ P(class|document) = \frac{P(document|class) \cdot P(class)}{P(document)} $$

In simpler terms:
- **P(class|document)**: Probability of a class given the document (what we want to find)
- **P(document|class)**: Probability of seeing this document if it belongs to this class
- **P(class)**: Prior probability of the class (how common it is in our training data)
- **P(document)**: Probability of the document (same for all classes, so we can ignore it for comparison)

**The "Naive" Assumption:**

The algorithm assumes that all features (words) are **conditionally independent** given the class. This means it assumes the presence of one word doesn't affect the probability of another word appearing. For example, it treats "hack" and "system" as independent, even though they often appear together.

Mathematically, this simplifies to:
$$ P(document|class) = P(word_1|class) \cdot P(word_2|class) \cdot ... \cdot P(word_n|class) $$

While this assumption is clearly false in natural language (words are related!), it makes the computation tractable and, surprisingly, still works quite well in practice!

**Why Naive Bayes is Great for Text Classification:**

- **Handles high-dimensional data**: Works well even when you have thousands of features (words)
- **Fast**: Training and prediction are very efficient - just counting occurrences!
- **Few parameters**: Only needs to estimate probabilities for each word in each class
- **Robust**: Less prone to overfitting compared to more complex models
- **Interpretable**: Easy to see which words contribute most to each prediction (as we'll explore later)
